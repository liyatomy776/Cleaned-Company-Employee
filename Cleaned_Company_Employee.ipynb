## 2. Cleaning Steps

- **drop_duplicates()** is utilized to eliminate exact duplicates in the rows.
- There is standardization of Department, Gender, City, and Work_Mode formats.
- **Joining_Date** is converted from an object/string to Pandas datetime format.
- For numeric variables, missing values are filled with the median since median value is less affected by phenomena like high salaries, ages, experience, or scores.
- For categorical variables, missing values are filled with a mode as they are nominal in nature.
- **dropna()** method is not applied for the main cleaning purpose as deletion of rows means loss of useful employee records.
- Forward Filling is not applied since all employee information is independent records; thus carrying the value of one employee over another employee is not acceptable.

```python
cleaned = df.copy()
cleaned = cleaned.drop_duplicates()
cleaned["Department"] = cleaned["Department"].replace({
    "ENGINEERING": "Engineering",
    "engineering": "Engineering"
})
cleaned["Gender"] = cleaned["Gender"].replace({
    "MALE": "Male",
    "female": "Female"
})
cleaned["City"] = cleaned["City"].astype("string").str.strip()
cleaned["City"] = cleaned["City"].replace({
    "DELHI": "Delhi",
    "delhi": "Delhi"
})
cleaned["Work_Mode"] = cleaned["Work_Mode"].replace({
    "REMOTE": "Remote",
    "remote": "Remote"
})
cleaned["Joining_Date"] = pd.to_datetime(
    cleaned["Joining_Date"],
    errors="coerce"
)
numeric_cols = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]
for col in numeric_cols:
    cleaned[col] = cleaned[col].fillna(cleaned[col].median())
categorical_cols = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]
for col in categorical_cols:
    cleaned[col] = cleaned[col].fillna(cleaned[col].mode()[0])
cleaned["Age"] = cleaned["Age"].astype(int)
cleaned["Performance_Score"] = cleaned["Performance_Score"].astype(int)
display(cleaned.head())
```

```python
print("Cleaned Shape:", cleaned.shape)

print("\nMissing Values After Cleaning:")
print(cleaned.isnull().sum())

print("\nDuplicate Records After Cleaning:", cleaned.duplicated().sum())

print("\nData Types After Cleaning:")
print(cleaned.dtypes)
```

```python
print("Department values after cleaning:")
print(sorted(cleaned["Department"].unique()))

print("\nGender values after cleaning:")
print(sorted(cleaned["Gender"].unique()))

print("\nCity values after cleaning:")
print(sorted(cleaned["City"].unique()))

print("\nWork_Mode values after cleaning:")
print(sorted(cleaned["Work_Mode"].unique()))
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Cells",
        "Duplicate Rows"
    ],
    "Before Cleaning": [
        df.shape[0],
        df.shape[1],
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum())
    ],
    "After Cleaning": [
        cleaned.shape[0],
        cleaned.shape[1],
        int(cleaned.isnull().sum().sum()),
        int(cleaned.duplicated().sum())
    ]
})

display(comparison)
```
